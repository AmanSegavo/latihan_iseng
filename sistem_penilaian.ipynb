```python
data = []
LABEL = {
    'label_Program_MBG' : "SISTEM PENILAIAN ALA SI MAHA TINGGI INDONESIA",
    'hei_tambahkan_data': "INPUT DATA",
    'HEIIIIII_tampilkan_Data' : "TAMPILAN DATA",
    'AKU_SUKA_SAWIT' : "NILAI TERTINGGI SAAT INI",
    '20JUTA_LAPANGAN_KERJA' : "PENGECEKAN LULUS"
}
bingkai = ("=" * len(LABEL['label_Program_MBG']))
```

```python
def tambah_data():

    # ini input nama pastinya berformat string bukan integer
    print("Masukkan nama siswa:", end=" ") # ini namanya label 
    nama_siswa = input()
    print(nama_siswa)
    # cek_tipe_data_siswa = type(nama_siswa)
    
    # print(cek_tipe_data_siswa)

    print("Masukkan nilai siswa: ", end="")
    # sekarang input nilai pastinya inns phones into surveillance detectorst'
    while True: # Jadi Pertanyaan Kalau semua function punya while True sendiri, program bisa berubah menjadi “negara dengan banyak pemerintahan yang tidak mau bubar” 😂
        try:
            nilai_siswa = int(input())
            if nilai_siswa > 100 or nilai_siswa < 0:
                print("Sorry ye nilai itu 0 sampe 100 yeeee")
            else:
                print(nilai_siswa)
                break
        except ValueError:
            print("Masuki lagi jangan typo\n")


    # cek_tipe_data_nilai = type(nilai_siswa)
    # print(cek_tipe_data_nilai
    data_satuan = [nama_siswa, nilai_siswa]
    # if len(data_satuan) == 0:
    #     print("hei datanya gak masuk")
    # else:
    #     print("Datanya masuk loh")
    data.append(data_satuan)
    if len(data) == 0:
        print("datanya gak masuk di penyimpanan utama")
    else:
        print("Ok data masuk : ", data)   

    return

def tampil_data():
    if len(data) == 0:
        print("Sorry Datanya belum ada...")
    else:
        count = 0
        for datum in data:

            count = count + 1 
            print(count,'.', datum)
            
    print
def kondisi():
    for datum in data:
        lulus = 75 # aturan ketat lulus Hacktiv8 hehe tapi harus sih ;D
        if datum[1] >= lulus:
            print("lulus")
        else:
            print("gak lulus")
    return 
        

def cari_nilai_tertinggi():
    if len(data) == 0:
        print("Kayaknya sekolah belum buka...")
        return

    nilai_terbaik = data[0]
    ada_kandidat_baru = False

    for data_baru in data[1:]:
        if data_baru[1] > nilai_terbaik[1]:
            nilai_terbaik = data_baru
            ada_kandidat_baru = True

    if ada_kandidat_baru:
        print("Ada kandidat yang mengalahkan kandidat sebelumnya.")

    else:
        print("Belum ada kandidat lebih bagus dari dia alias anggaran sekolah dipangkas hahahaha")

    print(f"Nilai tertinggi: {nilai_terbaik}")
    
```

```python
if len(data) == 0:
    print("Sorry ye datanya gak data ;D")
else:
    cari_nilai_tertinggi()
    print("Antek Antek Aseng")
```

```python


bebas_jadi_wni = 'exit'

def kondisi_rakyat():
    while True:

        print("Masukkan Pilihan")
        rakyat_kecil = input()
        print(bingkai)
        print(LABEL['label_Program_MBG'])
        print(bingkai)
        if rakyat_kecil.lower() == bebas_jadi_wni:
            print("YES BEBAS DARI PROGRAM MBG")
            break

        try:
            pilihan = int(rakyat_kecil)
        except ValueError:
            print("Masukkan pilihan yang benar.")
            continue
        if pilihan == 1:
    
                print(bingkai)
                print(LABEL['hei_tambahkan_data'])
                print(bingkai)
                tambah_data()

        elif pilihan == 2:
            print(bingkai)
            print(LABEL['HEIIIIII_tampilkan_Data'])
            print(bingkai)
            tampil_data()
        elif pilihan == 3:
            print(bingkai)
            print(LABEL['20JUTA_LAPANGAN_KERJA'])
            print(bingkai)
            kondisi()
        elif pilihan == 4:
            print(bingkai)
            print(LABEL['AKU_SUKA_SAWIT'])
            print(bingkai)
            cari_nilai_tertinggi()
        else:
            print("Masukkin yang bener ahhh")

kondisi_rakyat()
```
